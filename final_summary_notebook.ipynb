```python
import pandas as pd

from pathlib import Path
```

```python
BASE_DIR = Path(
    "paper_implementation"
)

RESULTS_DIR = (
    BASE_DIR /
    "results"
)
```

```python
modes = [

    "mode_1",
    "mode_2",
    "mode_3",
    "mode_4",
    "mode_5"
]
```

```python

summary_rows = []

for mode in modes:

    metrics_file = (
        RESULTS_DIR /
        mode /
        "daily_metrics.csv"
    )

    # ----------------------------------------
    # Check file existence
    # ----------------------------------------

    if not metrics_file.exists():

        print()

        print(
            f"Missing metrics file for {mode}"
        )

        print()

        continue

    df = pd.read_csv(
        metrics_file
    )

    # ----------------------------------------
    # Validate completed days
    # ----------------------------------------

    unique_days = (
        df["Day"]
        .nunique()
    )

    if unique_days != 31:

        print()

        print(
            f"WARNING: {mode} contains "
            f"{unique_days} days instead of 31"
        )

        print()

    # ----------------------------------------
    # Mean Metrics
    # ----------------------------------------

    mean_sklearn_auc = (
        df["Sklearn_AUC"]
        .mean()
    )

    mean_paper_auc = (
        df["Paper_AUC"]
        .mean()
    )

    mean_map10 = (
        df["MAP@10"]
        .mean()
    )

    mean_dbcv = (
        df["DBCV"]
        .mean()
    )

    mean_noise = (
        df["Noise_Consumers"]
        .mean()
    )

    # ----------------------------------------
    # Standard Deviations
    # ----------------------------------------

    std_sklearn_auc = (
        df["Sklearn_AUC"]
        .std()
    )

    std_paper_auc = (
        df["Paper_AUC"]
        .std()
    )

    std_map10 = (
        df["MAP@10"]
        .std()
    )

    # ----------------------------------------
    # Summary Row
    # ----------------------------------------

    summary_rows.append({

        "Mode":
        mode,

        "Completed_Days":
        unique_days,

        "Mean_Sklearn_AUC":
        mean_sklearn_auc,

        "STD_Sklearn_AUC":
        std_sklearn_auc,

        "Mean_Paper_AUC":
        mean_paper_auc,

        "STD_Paper_AUC":
        std_paper_auc,

        "Mean_MAP@10":
        mean_map10,

        "STD_MAP@10":
        std_map10,

        "Mean_DBCV":
        mean_dbcv,

        "Mean_Noise_Consumers":
        mean_noise
    })
```

```python
summary_df = pd.DataFrame(
    summary_rows
)

summary_df = (
    summary_df
    .sort_values(
        by="Mode"
    )
    .reset_index(drop=True)
)
```

```python
print()

print(
    "FINAL TABLE III STYLE SUMMARY"
)

print()

display(
    summary_df
)
```

```python
output_file = (
    RESULTS_DIR /
    "table3_summary.csv"
)

summary_df.to_csv(
    output_file,
    index=False
)

print()

print(
    "Summary table saved."
)

print()

print(
    output_file
)
```
