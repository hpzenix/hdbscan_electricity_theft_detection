```python
import pandas as pd
import numpy as np

from pathlib import Path
```

```python
BASE_DIR = Path(
    "paper_implementation"
)

THEFT_DIR = (
    BASE_DIR /
    "theft_simulation"
)

PREPROCESS_DIR = (
    BASE_DIR /
    "preprocessing"
)

MISSING_DIR = (
    PREPROCESS_DIR /
    "missing_value_recovery"
)

MISSING_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
modes = [
    "mode_1",
    "mode_2",
    "mode_3",
    "mode_4",
    "mode_5"
]

for mode in modes:

    (
        MISSING_DIR /
        mode
    ).mkdir(
        parents=True,
        exist_ok=True
    )

print(
    "Folders created successfully."
)
```

```python
hour_cols = [
    f"HOUR_{i}"
    for i in range(24)
]
```

```python
def recover_missing_values(df):

    recovered_count = 0

    for row_idx in df.index:

        day_vector = (
            df.loc[
                row_idx,
                hour_cols
            ]
            .astype(float)
        )

        # Edge case:
        # all 24 values are NaN

        if day_vector.isna().all():

            continue

        daily_mean = (
            day_vector.mean()
        )

        for hour in hour_cols:

            value = df.loc[
                row_idx,
                hour
            ]

            if pd.isna(value):

                df.loc[
                    row_idx,
                    hour
                ] = daily_mean

                recovered_count += 1

    return df, recovered_count
```

```python
MODE = "mode_1"

INPUT_DIR = (
    THEFT_DIR /
    MODE
)

OUTPUT_DIR = (
    MISSING_DIR /
    MODE
)

files = sorted(
    INPUT_DIR.glob("*.csv")
)

total_recovered = 0

for file in files:

    df = pd.read_csv(file)

    df, recovered = recover_missing_values(df)

    total_recovered += recovered

    df.to_csv(
        OUTPUT_DIR /
        file.name,
        index=False
    )

print(
    "Mode 1 Missing Values Recovered:",
    total_recovered
)
```

```python
remaining_nans = 0

for file in sorted(
    (
        MISSING_DIR /
        "mode_1"
    ).glob("*.csv")
):

    df = pd.read_csv(file)

    df[hour_cols] = df[hour_cols].astype(float)

    remaining_nans += (
        df[hour_cols]
        .isna()
        .sum()
        .sum()
    )

print(
    "Remaining NaNs in Mode 1:",
    remaining_nans
)
```
