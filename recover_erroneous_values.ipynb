```python
import pandas as pd
import numpy as np

from pathlib import Path
```

```python
BASE_DIR = Path(
    "paper_implementation"
)

PREPROCESS_DIR = (
    BASE_DIR /
    "preprocessing"
)

INPUT_BASE_DIR = (
    PREPROCESS_DIR /
    "missing_value_recovery"
)

ERROR_DIR = (
    PREPROCESS_DIR /
    "erroneous_value_recovery"
)

ERROR_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
modes = [
    "mode_1",
    "mode_2",
    "mode_3",
    "mode_4",
    "mode_5"
]

for mode in modes:

    (
        ERROR_DIR /
        mode
    ).mkdir(
        parents=True,
        exist_ok=True
    )

print(
    "Folders created successfully."
)
```

```python
hour_cols = [
    f"HOUR_{i}"
    for i in range(24)
]
```

```python
def recover_erroneous_values(df):

    corrected_count = 0

    total_days = len(df)

    for row_idx in df.index:

        # Current day's vector

        day_vector = (
            df.loc[
                row_idx,
                hour_cols
            ]
            .astype(float)
            .values
        )

        # Daily sigma

        sigma = np.std(
            day_vector
        )

        threshold = (
            3 * sigma
        )

        corrected_vector = (
            day_vector.copy()
        )

        for t in range(24):

            current_value = (
                day_vector[t]
            )

            # ------------------------------------------------
            # Ignore:
            # First reading of Day 1
            # Last reading of Day 31
            # ------------------------------------------------

            if (
                row_idx == 0
                and t == 0
            ):

                continue

            if (
                row_idx == total_days - 1
                and t == 23
            ):

                continue

            # ------------------------------------------------
            # Equation (5)
            # ------------------------------------------------

            if current_value > threshold:

                # --------------------------------------------
                # Previous value
                # --------------------------------------------

                if t == 0:

                    # Previous day's HOUR_23

                    prev_value = float(
                        df.loc[
                            row_idx - 1,
                            "HOUR_23"
                        ]
                    )

                else:

                    prev_value = (
                        day_vector[t - 1]
                    )

                # --------------------------------------------
                # Next value
                # --------------------------------------------

                if t == 23:

                    # Next day's HOUR_0

                    next_value = float(
                        df.loc[
                            row_idx + 1,
                            "HOUR_0"
                        ]
                    )

                else:

                    next_value = (
                        day_vector[t + 1]
                    )

                # --------------------------------------------
                # Replacement
                # --------------------------------------------

                replacement = (
                    prev_value
                    +
                    next_value
                ) / 2

                corrected_vector[t] = (
                    replacement
                )

                corrected_count += 1

        # Save corrected vector

        df.loc[
            row_idx,
            hour_cols
        ] = corrected_vector

    return df, corrected_count
```

```python
MODE = "mode_1"

INPUT_DIR = (
    INPUT_BASE_DIR /
    MODE
)

OUTPUT_DIR = (
    ERROR_DIR /
    MODE
)

files = sorted(
    INPUT_DIR.glob("*.csv")
)

total_corrected = 0

for file in files:

    df = pd.read_csv(file)

    df, corrected = (
        recover_erroneous_values(df)
    )

    total_corrected += corrected

    df.to_csv(
        OUTPUT_DIR /
        file.name,
        index=False
    )

print(
    "Mode 1 Erroneous Values Corrected:",
    total_corrected
)
```

```python
remaining_outliers = 0

for file in sorted(
    (
        ERROR_DIR /
        "mode_1"
    ).glob("*.csv")
):

    df = pd.read_csv(file)

    df[hour_cols] = (
        df[hour_cols]
        .astype(float)
    )

    for row_idx in df.index:

        vector = (
            df.loc[
                row_idx,
                hour_cols
            ]
            .values
        )

        sigma = np.std(vector)

        threshold = (
            3 * sigma
        )

        remaining_outliers += (
            vector > threshold
        ).sum()

print(
    "Remaining Outliers in Mode 1:",
    remaining_outliers
)
```

```python
# ==========================================
# SELECT MODE AND METER
# ==========================================

MODE = "mode_4"

METER_ID = "17374"


# ==========================================
# PATHS
# ==========================================

before_path = (
    INPUT_BASE_DIR /
    MODE /
    f"{METER_ID}.csv"
)

after_path = (
    ERROR_DIR /
    MODE /
    f"{METER_ID}.csv"
)


# ==========================================
# LOAD FILES
# ==========================================

before_df = pd.read_csv(
    before_path
)

after_df = pd.read_csv(
    after_path
)


# ==========================================
# FIND DIFFERENCES
# ==========================================

difference_records = []

for row_idx in before_df.index:

    date = before_df.loc[
        row_idx,
        "DATE"
    ]

    for hour in hour_cols:

        before_value = float(
            before_df.loc[
                row_idx,
                hour
            ]
        )

        after_value = float(
            after_df.loc[
                row_idx,
                hour
            ]
        )

        if before_value != after_value:

            difference_records.append({

                "Date":
                date,

                "Hour":
                hour,

                "Before":
                before_value,

                "After":
                after_value,

                "Difference":
                after_value - before_value
            })


# ==========================================
# CREATE COMPARISON TABLE
# ==========================================

comparison_df = pd.DataFrame(
    difference_records
)


# ==========================================
# RESULTS
# ==========================================

print(
    "Total Modified Readings:",
    len(comparison_df)
)

comparison_df.head(50)
```

```python
MODE = "mode_3"
METER_ID = "78"

file_path = (
    INPUT_BASE_DIR /
    MODE /
    f"{METER_ID}.csv"
)

df = pd.read_csv(file_path)

results = []

for row_idx in df.index:

    date = df.loc[row_idx, "DATE"]

    vector = (
        df.loc[row_idx, hour_cols]
        .astype(float)
        .values
    )

    sigma = np.std(vector)

    threshold = 3 * sigma

    hour0 = vector[0]
    hour23 = vector[23]

    results.append({

        "Date": date,

        "Threshold": threshold,

        "HOUR_0": hour0,

        "HOUR_0 > Threshold":
        hour0 > threshold,

        "HOUR_23": hour23,

        "HOUR_23 > Threshold":
        hour23 > threshold
    })

boundary_df = pd.DataFrame(results)

boundary_df
```
