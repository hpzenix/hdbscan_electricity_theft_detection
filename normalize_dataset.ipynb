```python
import pandas as pd
import numpy as np

from pathlib import Path
```

```python
BASE_DIR = Path(
    "paper_implementation"
)

PREPROCESS_DIR = (
    BASE_DIR /
    "preprocessing"
)

INPUT_BASE_DIR = (
    PREPROCESS_DIR /
    "erroneous_value_recovery"
)

NORMALIZED_DIR = (
    PREPROCESS_DIR /
    "normalized_data"
)

NORMALIZED_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
modes = [
    "mode_1",
    "mode_2",
    "mode_3",
    "mode_4",
    "mode_5"
]

for mode in modes:

    (
        NORMALIZED_DIR /
        mode
    ).mkdir(
        parents=True,
        exist_ok=True
    )

print(
    "Normalization folders created successfully."
)
```

```python
hour_cols = [
    f"HOUR_{i}"
    for i in range(24)
]
```

```python
def normalize_dataset(df):

    for row_idx in df.index:

        # Daily load vector

        day_vector = (
            df.loc[
                row_idx,
                hour_cols
            ]
            .astype(float)
            .values
        )

        # Maximum value of current day

        max_value = np.max(
            day_vector
        )

        # ----------------------------------------
        # If entire vector is zero,
        # keep it unchanged
        # ----------------------------------------

        if max_value == 0:

            normalized_vector = (
                day_vector.copy()
            )

        else:

            normalized_vector = (
                day_vector / max_value
            )

        # Save normalized values

        df.loc[
            row_idx,
            hour_cols
        ] = normalized_vector

    return df
```

```python
MODE = "mode_1"

INPUT_DIR = (
    INPUT_BASE_DIR /
    MODE
)

OUTPUT_DIR = (
    NORMALIZED_DIR /
    MODE
)

files = sorted(
    INPUT_DIR.glob("*.csv")
)

for file in files:

    df = pd.read_csv(file)

    df = normalize_dataset(df)

    df.to_csv(
        OUTPUT_DIR /
        file.name,
        index=False
    )

print(
    "Mode 1 normalization completed."
)
```

```python
all_valid = True

for file in sorted(
    (
        NORMALIZED_DIR /
        "mode_1"
    ).glob("*.csv")
):

    df = pd.read_csv(file)

    df[hour_cols] = (
        df[hour_cols]
        .astype(float)
    )

    for row_idx in df.index:

        vector = (
            df.loc[
                row_idx,
                hour_cols
            ]
            .values
        )

        max_value = np.max(vector)

        if not np.isclose(max_value, 1.0):

            if max_value != 0:

                all_valid = False

                print(
                    file.name,
                    row_idx,
                    max_value
                )

print(
    "Mode 1 normalization valid:",
    all_valid
)
```
