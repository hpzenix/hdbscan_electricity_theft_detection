```python
import pandas as pd
import numpy as np

from pathlib import Path

np.random.seed(42)
```

```python
DATA_DIR = Path("daily_profiles_24h")

all_files = sorted(DATA_DIR.glob("*.csv"))

print("Total meters:", len(all_files))
```

```python
BASE_DIR = Path("paper_implementation")

(BASE_DIR / "selected_150_meters").mkdir(
    parents=True,
    exist_ok=True
)

(BASE_DIR / "area_assignments").mkdir(
    parents=True,
    exist_ok=True
)
```

```python
meter_ids = [
    f.stem
    for f in all_files
]

selected_150 = np.random.choice(
    meter_ids,
    size=150,
    replace=False
)

print(
    "Selected meters:",
    len(selected_150)
)
```

```python
selected_df = pd.DataFrame({
    "Meter": selected_150
})

selected_df.to_csv(
    BASE_DIR /
    "selected_150_meters" /
    "selected_150_meter_ids.csv",
    index=False
)

selected_df.head()
```

```python
selected_150 = list(selected_150)

np.random.shuffle(selected_150)

area_1 = selected_150[:50]
area_2 = selected_150[50:100]
area_3 = selected_150[100:150]

print(len(area_1))
print(len(area_2))
print(len(area_3))
```

```python
records = []

for meter in area_1:
    records.append([
        meter,
        "Area_1"
    ])

for meter in area_2:
    records.append([
        meter,
        "Area_2"
    ])

for meter in area_3:
    records.append([
        meter,
        "Area_3"
    ])

area_df = pd.DataFrame(
    records,
    columns=[
        "Meter",
        "Area"
    ]
)

area_df.to_csv(
    BASE_DIR /
    "area_assignments" /
    "meter_area_mapping.csv",
    index=False
)

area_df.head()
```

```python
fraud_area1 = np.random.choice(
    area_1,
    9,
    replace=False
)

fraud_area2 = np.random.choice(
    area_2,
    9,
    replace=False
)

fraud_area3 = np.random.choice(
    area_3,
    9,
    replace=False
)
```

```python
fraud_df = pd.DataFrame({

    "Meter":
        list(fraud_area1)
        + list(fraud_area2)
        + list(fraud_area3),

    "Area":
        ["Area_1"] * 9
        + ["Area_2"] * 9
        + ["Area_3"] * 9
})

fraud_df.to_csv(
    BASE_DIR /
    "area_assignments" /
    "fraud_consumers.csv",
    index=False
)

fraud_df
```

```python
print(
    "Total Selected Consumers:",
    len(selected_150)
)

print(
    "Area 1:",
    len(area_1)
)

print(
    "Area 2:",
    len(area_2)
)

print(
    "Area 3:",
    len(area_3)
)

print(
    "Fraud Consumers:",
    len(fraud_df)
)
```
